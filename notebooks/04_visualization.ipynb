```python
# Single plot for all

banks = ["BAC", "C", "GS", "JPM", "WFC"]

plt.figure(figsize=(10,6))

for bank in banks:
    df = pd.read_csv(f"../data/processed/{bank}_balance_sheet_clean.csv")
    plt.plot(df.index, df["Total Assets"], label=bank)

plt.title("Total Assets Over Time - Major US Banks")
plt.xlabel("Year")
plt.ylabel("Total Assets")
plt.legend()
plt.grid(False)

plt.tight_layout()
plt.savefig("../outputs/figures/all_banks_assets_trend.png")
plt.show()
```
