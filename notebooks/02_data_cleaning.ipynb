## Cleaning notebook

```python
import pandas as pd
df = pd.read_csv("../data/raw/JPM_balance_sheet.csv", index_col=0)
df.head()
```

```python
df=df.T
```

```python
# Deal with nulls
df.isna
df = df.dropna(axis=1, how="all")
df = df.fillna(0)

```

```python
# Scale values
df = df / 1_000_000
df.head()
```

```python
df.to_csv("../data/processed/JPM_balance_sheet_clean.csv")
```
