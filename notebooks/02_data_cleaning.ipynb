## Cleaning notebook

```python
import pandas as pd
df = pd.read_csv("../data/raw/JPM_balance_sheet.csv", index_col=0)
df.head()
```

**Clean the income statements and Balance sheet**

**Single code for all cleaned balance sheet statements**

```python
tickers = ["JPM", "BAC", "GS", "WFC", "C"]

for ticker in tickers:
    # Read raw file
    df = pd.read_csv(f"../data/raw/{ticker}_balance_sheet.csv",index_col=0)

    # Cleaning steps
    df = df.T
    df = df.dropna(axis=1, how="all")
    df = df.fillna(0)
    df = (df / 1_000_000).round(2)

    # Save cleaned file
    df.to_csv(f"../data/processed/{ticker}_balance_sheet_clean.csv")
```

**Single code for all Cleaned Income statements updates**

```python
tickers = ["JPM", "BAC", "GS", "WFC", "C"]

for ticker in tickers:
    # Read raw file
    df = pd.read_csv(f"../data/raw/{ticker}_income_statement.csv",index_col=0)

    # Cleaning steps
    df = df.T
    df = df.dropna(axis=1, how="all")
    df = df.fillna(0)
    df = (df / 1_000_000).round(2)

    # Save cleaned file
    df.to_csv(f"../data/processed/{ticker}_income_statement_clean.csv")
```

**Cashflow Statements Cleaned data**

```python
tickers = ["JPM", "BAC", "GS", "WFC", "C"]

for ticker in tickers:
    df=pd.read_csv(f"../data/raw/{ticker}_cash_flows.csv",index_col=0)

    # Cleaning steps
    df = df.T
    df = df.dropna(axis=1, how="all")
    df = df.fillna(0)
    df = (df / 1_000_000).round(2)

    # Save cleaned file
    df.to_csv(f"../data/processed/{ticker}_cash_flows_clean.csv")
```
