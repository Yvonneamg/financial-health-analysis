## Cleaning notebook

```python
import pandas as pd
df = pd.read_csv("../data/raw/JPM_balance_sheet.csv", index_col=0)
df.head()
```

```python
df.index
```

```python
df=df.T
```

```python
# Deal with nulls
df.isna
df = df.dropna(axis=1, how="all")
df = df.fillna(0)

```

```python
# Scale values
df = (df / 1_000_000).round(2)
df.head()
```

```python
df.to_csv("../data/processed/JPM_balance_sheet_clean.csv")
```

**Clean the income statements**

```python
df=pd.read_csv("../data/raw/JPM_income_statement.csv",index_col=0)
df=df.T
df=df.dropna(axis=1, how='all')
df=df.fillna(0)
df=(df/1_000_000).round(2)
df.head()
df.to_csv('../data/processed/JPM_income_statement_clean.csv')

```

**Single code for all Cleaned Income statements updates**

```python
tickers = ["JPM", "BAC", "GS", "WFC", "C"]

for ticker in tickers:
    # Read raw file
    df = pd.read_csv(f"../data/raw/{ticker}_income_statement.csv",index_col=0)

    # Cleaning steps
    df = df.T
    df = df.dropna(axis=1, how="all")
    df = df.fillna(0)
    df = (df / 1_000_000).round(2)

    # Save cleaned file
    df.to_csv(f"../data/processed/{ticker}_income_statement_clean.csv")
```
